# Python Login Access Variables

A beginner practice notebook based on a security-analyst login scenario.

**Learning goals:** assign variables, inspect data types with `type()`, update a list, and evaluate a Boolean comparison.

> Learning exercise only: this notebook does not authenticate users or secure a real device. The usernames are sample data.

## Task 1 — Assign the device ID
Store the device identifier as a string, then display it.

```python
device_id = "72e08x0"
print(device_id)
```

## Task 2 — Check the device ID data type
Use `type()` to inspect the variable and store the result in `device_id_type`.

**Observation:** `device_id` is a `str` (string) because it is text enclosed in quotation marks.

```python
device_id = "72e08x0"
device_id_type = type(device_id)
print(device_id_type)
```

## Task 3 — Create the approved username list
A Python list stores multiple values in one variable.

**Observation:** `username_list` has the data type `list`.

```python
username_list = ["madebowa", "jnguyen", "tbecker", "nhersh", "redwards"]
print(username_list)
```

## Task 5 — Update the username list
Reassign the variable with the updated list, including `lpope`. Reassignment replaces the previous list value.

**Observation:** the updated list contains six usernames instead of five.

```python
username_list = ["madebowa", "jnguyen", "tbecker", "nhersh", "redwards"]
print(username_list)

username_list = ["madebowa", "jnguyen", "tbecker", "nhersh", "redwards", "lpope"]
print(username_list)
```

## Task 6 — Set the maximum login attempts
**Observation:** `max_logins` has the data type `int` (integer), because `3` is a whole number.

```python
max_logins = 3
max_logins_type = type(max_logins)
print(max_logins_type)
```

## Task 7 — Track current login attempts
**Observation:** `login_attempts` has the data type `int` (integer), because `2` is a whole number.

```python
login_attempts = 2
login_attempts_type = type(login_attempts)
print(login_attempts_type)
```

## Task 8 — Compare attempts with the limit
The expression `login_attempts <= max_logins` checks whether the current number of attempts is less than or equal to the maximum. With 2 attempts and a limit of 3, the result is `True`.

```python
max_logins = 3
login_attempts = 2
print(login_attempts <= max_logins)
```

## Task 9 — Test a value above the limit
With 4 attempts and a maximum of 3, the comparison returns `False`. The comparison itself does not block a login; it only evaluates to a Boolean value.

```python
max_logins = 3
login_attempts = 4
print(login_attempts <= max_logins)
```

## Task 10 — Set the login status
**Observation:** `login_status` has the data type `bool` (Boolean). A Boolean value is either `True` or `False`.

```python
login_status = False
login_status_type = type(login_status)
print(login_status_type)
```
